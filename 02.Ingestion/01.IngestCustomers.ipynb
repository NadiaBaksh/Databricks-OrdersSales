#### READ `CUSTOMERS` FROM VOLUME

```python
BASE_DIR = '/Volumes/sales/landing/sales/Customers/'
CATALOG = 'SALES'
SCHEMA = 'BRONZE'
TARGET_TABLE = 'CUSTOMERS'
```

```python
cust_schema = '''customer_id STRING, customer_unique_id STRING, 
              customer_zip_code_prefix INT, 
              customer_city STRING, 
              customer_state STRING'''
```

```python
cust_df = spark.read.format('csv') \
                    .option('header', True) \
                    .schema(cust_schema) \
                    .load(f'{BASE_DIR}')
display(cust_df.limit(3))
# cust_df.printSchema()
```

```python
from pyspark.sql.functions import *

cust_trans_df = cust_df.select('*', "_metadata.file_name", "_metadata.file_path", 
                               "_metadata.file_modification_time", current_timestamp().alias('ingestion_timestamp')
)
display(cust_trans_df.limit(1))
```

```python
cust_trans_df.write.mode('append').format('delta').saveAsTable(f'{CATALOG}.{SCHEMA}.{TARGET_TABLE}')
```
